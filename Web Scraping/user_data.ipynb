```python
#converting json to csv

import json
import csv

# Input JSON file and output CSV file
input_file = 'C:/Users/dell/Desktop/major_project/project-final/Tourist-Attraction-Recommendation-System/Web Scraping/outputs/reviews_valid_urls1.json'
# Replace with your JSON file name
output_file = 'reviews_valid_urls1.csv'  # Desired output CSV file name

# Load JSON data
with open(input_file, 'r', encoding='utf-8') as json_file:
    data = json.load(json_file)

# Ensure the data is a list of dictionaries
if not isinstance(data, list):
    data = [data]

# Extract headers from keys
headers = data[0].keys()

# Write to CSV
with open(output_file, 'w', encoding='utf-8', newline='') as csv_file:
    writer = csv.DictWriter(csv_file, fieldnames=headers)
    writer.writeheader()

    # Write rows, handling nested structures (like photos) by converting them to strings
    for row in data:
        row['photos'] = ", ".join(row['photos']) if isinstance(row['photos'], list) else row['photos']
        writer.writerow(row)

print(f"JSON data successfully converted to {output_file}")
```

```python
import pandas as pd

# List of file paths for the CSV files
file_paths = ['C:/Users/dell/Desktop/major_project/project-final/Tourist-Attraction-Recommendation-System/Web Scraping/outputs/userReview_data/reviews_valid_urls1.csv', 'C:/Users/dell/Desktop/major_project/project-final/Tourist-Attraction-Recommendation-System/Web Scraping/outputs/userReview_data/reviews_valid_urls.csv', 'C:/Users/dell/Desktop/major_project/project-final/Tourist-Attraction-Recommendation-System/Web Scraping/outputs/userReview_data/reviews_attraction_nepal.csv', 'C:/Users/dell/Desktop/major_project/project-final/Tourist-Attraction-Recommendation-System/Web Scraping/outputs/userReview_data/reviews_attraction_nepal1.csv']

# Read and combine all CSV files into a single DataFrame
combined_df = pd.concat([pd.read_csv(file) for file in file_paths], ignore_index=True)

# Add a unique 'reviewer_id' column
# Example: Using the index (1, 2, 3,...)
combined_df['reviewer_id'] = combined_df.index + 1

# Save the combined DataFrame to a new CSV file
combined_df.to_csv('combined_with_reviewer_id.csv', index=False)

print("All files have been combined into 'combined_with_reviewer_id.csv' with reviewer_id added.")
```

```python
import pandas as pd

# Load the existing file
file_path = 'C:/Users/dell/Desktop/major_project/project-final/Tourist-Attraction-Recommendation-System/Web Scraping/outputs/userReview_data/combined_with_reviewer_id.csv'
df = pd.read_csv(file_path)

# Rename columns
df.rename(columns={

    'reviewer_name': 'user_name',
    'reviewer_id': 'user_id',
    'text': 'review_text'
}, inplace=True)

# Rearrange columns in the desired order
rearranged_columns = ['user_id', 'user_name','place_name', 'rating', 'review_text', 'date', 'likes_count', 
                      'reviewer_link', 'reviewer_image', 'photos']
df = df[rearranged_columns]

# Save the updated DataFrame to a new CSV file
output_file = 'user_info.csv'
df.to_csv(output_file, index=False)

print(f"Columns have been renamed and rearranged. The updated file is saved as '{output_file}'.")

```
