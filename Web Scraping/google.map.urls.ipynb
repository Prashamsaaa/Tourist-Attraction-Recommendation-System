```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import time
import csv
import os

# Initialize WebDriver
options = webdriver.ChromeOptions()
options.add_argument("start-maximized")
options.add_argument("--disable-blink-features=AutomationControlled")  # Reduce automation detection
options.add_argument("--lang=en")  # Force English language

# Initialize WebDriver
driver = webdriver.Chrome(options=options)

# Input and output file paths
input_file = "../Data/FinalDataset/Data.csv"  # Input CSV with places
output_file = "data_url.csv"               # Output CSV to save the URLs

# Check if input file exists
if not os.path.exists(input_file):
    print(f"Input file not found: {input_file}")
    driver.quit()
    exit()

# List to hold place names and their URLs
places = []
urls = []

# Read the CSV and extract the place names
try:
    with open(input_file, mode="r", newline="", encoding="utf-8") as file:
        reader = csv.DictReader(file)
        for row in reader:
            place = row["name"].strip()  # Get and clean the place name
            places.append(place)
    print(f"Found {len(places)} places to process.")
except Exception as e:
    print(f"Error reading input file: {e}")
    driver.quit()
    exit()

# Process each place
for index, place in enumerate(places, start=1):
    try:
        # Open Google Maps
        driver.get("https://www.google.com/maps?hl=en")
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.ID, "searchboxinput"))
        )

        # Search for the place
        search_box = driver.find_element(By.ID, "searchboxinput")
        search_box.clear()
        search_box.send_keys(place)
        search_box.send_keys(Keys.RETURN)  # Press Enter

        # Wait for the first result and click it
        WebDriverWait(driver, 10).until(
            EC.element_to_be_clickable(
                (By.XPATH, '//*[@id="QA0Szd"]/div/div/div[1]/div[2]/div/div[1]/div/div/div[3]/div/div/button[2]/div[2]/div[2]')
            )
        ).click()
        print(f"[{index}/{len(places)}] Navigating to details for '{place}'")

        # Wait for the place page to load and capture the URL
        WebDriverWait(driver, 10).until(
            EC.url_contains("https://www.google.com/maps/place")
        )
        current_url = driver.current_url
        urls.append(current_url)
        print(f"Saved URL for '{place}': {current_url}")

    except Exception as e:
        print(f"Error processing '{place}': {e}")
        urls.append("Error")

# Quit the driver
driver.quit()

# Save results to the output CSV
try:
    with open(output_file, mode="w", newline="", encoding="utf-8") as file:
        writer = csv.writer(file)
        writer.writerow(["name", "URL"])  # Write header
        for place, url in zip(places, urls):
            writer.writerow([place, url])
    print(f"URLs saved to '{output_file}'")
except Exception as e:
    print(f"Error saving results: {e}")

# Save URLs to a text file for convenience (optional)
try:
    with open("google_maps_urls.txt", mode="w", encoding="utf-8") as file:
        for place, url in zip(places, urls):
            file.write(f"{place}: {url}\n")
    print("URLs also saved to 'google_maps_urls.txt'")
except Exception as e:
    print(f"Error saving text file: {e}")
```
