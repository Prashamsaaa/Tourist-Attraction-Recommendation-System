IMPORT NECESSARY LIBRARIES

```python
import pandas as pd
import json
from sklearn.preprocessing import MultiLabelBinarizer, LabelEncoder
```

LOADING DATASET

```python
data = pd.read_csv('Output/PreparedData.csv')
print("Dataset loaded successfully.")
```

```python
with open('../Data/FinalDataset/CategorizedTags.json', 'r') as f:
    categorized_tags = json.load(f)
print("Categorized tags loaded successfully.")
```

```python
province_encoder = LabelEncoder()
category_encoder = LabelEncoder()
mlb = MultiLabelBinarizer()
```

```python
encoded_provinces = province_encoder.fit_transform(data['Province'])
```

```python
def determine_category(tags, categorized_tags):
    for category, category_tags in categorized_tags.items():
        if any(tag in category_tags for tag in tags):
            return category
    return 'Uncategorized'
```

```python
encoded_categories = []
for tags in data['Tags']:
    category = determine_category(tags.split(), categorized_tags)
    encoded_categories.append(category)
```

```python
encoded_categories = category_encoder.fit_transform(encoded_categories)
```

```python
encoded_tags = mlb.fit_transform(data['Tags'].apply(lambda x: x.split()))
```

```python
encoded_data = pd.DataFrame(encoded_tags, columns=mlb.classes_)
encoded_data['Encoded_Province'] = encoded_provinces
encoded_data['Encoded_Category'] = encoded_categories
```

```python
final_data = pd.concat([data, encoded_data], axis=1)
print("Encoding completed successfully.")

# Save the encoded data to a new CSV file
final_data.to_csv('Output/EncodedData.csv', index=False)
print("Encoded data saved successfully to 'EncodedData.csv'.")
```
